```python
# ===============================================
# 📦 1. Extract the dataset and load model
# ===============================================
import zipfile, os

zip_path = '/content/DCPL-Detection-of-Covid-19-Pneumonia-And-Lung_Nodule-Cancer-Using-Deep-Learning-Technique-main.zip'
extract_dir = '/content/model/DCPL-Detection-of-Covid-19-Pneumonia-And-Lung_Nodule-Cancer-Using-Deep-Learning-Technique-main'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall('/content/model')
print("✅ Project extracted successfully.")

from keras.models import load_model
from tensorflow.keras.optimizers import Adam # Import an optimizer, e.g., Adam

classifier = load_model(f'{extract_dir}/DCPL.h5')

# Explicitly compile the model after loading to ensure metrics are set correctly
classifier.compile(optimizer=Adam(), loss='categorical_crossentropy', metrics=['accuracy'])


print("\n📘 Model Summary:")
classifier.summary()

# ===============================================
# 📁 2. Prepare test dataset
# ===============================================
from tensorflow.keras.preprocessing.image import ImageDataGenerator

test_dir = f'{extract_dir}/Sample_dataset/Test'
test_datagen = ImageDataGenerator(rescale=1./255)
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(64, 64),
    batch_size=16,
    class_mode='categorical',
    shuffle=False
)

# ===============================================
# 📈 3. Evaluate model
# ===============================================
loss, acc = classifier.evaluate(test_generator, verbose=1)
print(f"\n✅ Test Accuracy: {acc*100:.2f}%")
print(f"❌ Test Loss: {loss:.4f}")

# ===============================================
# 🧮 4. Confusion Matrix and Classification Report
# ===============================================
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

Y_pred = classifier.predict(test_generator)
y_pred = np.argmax(Y_pred, axis=1)
cm = confusion_matrix(test_generator.classes, y_pred)
labels = list(test_generator.class_indices.keys())

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

print("\n📊 Classification Report:")
print(classification_report(test_generator.classes, y_pred, target_names=labels))

# ===============================================
# 📉 5. Accuracy/Loss Graph (if history available)
# ===============================================
import json
history_path = f'{extract_dir}/history.json'

if os.path.exists(history_path):
    with open(history_path, 'r') as f:
        history = json.load(f)

    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1)
    plt.plot(history['accuracy'], label='Train Accuracy')
    plt.plot(history['val_accuracy'], label='Val Accuracy')
    plt.title('Model Accuracy')
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(history['loss'], label='Train Loss')
    plt.plot(history['val_loss'], label='Val Loss')
    plt.title('Model Loss')
    plt.legend()
    plt.show()
else:
    print("\n⚠️ No training history file found. Skipping accuracy/loss graph.")

# ===============================================
# 🔵 6. ROC Curve and AUC
# ===============================================
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize

# Binarize labels for ROC (multi-class)
y_true = test_generator.classes
y_true_bin = label_binarize(y_true, classes=list(range(len(labels))))

plt.figure(figsize=(8,6))

for i in range(len(labels)):
    fpr, tpr, _ = roc_curve(y_true_bin[:, i], Y_pred[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f'{labels[i]} (AUC = {roc_auc:.2f})')

plt.plot([0, 1], [0, 1], 'k--')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.show()
```

```python


import numpy as np
import cv2
import matplotlib.pyplot as plt
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing import image
import tensorflow as tf
from tensorflow.keras.layers import Conv2D, Input # Import Input to define new input tensor

# ---- Select any test image ----
img_path = input("📸 Enter the absolute path of an image for Grad-CAM visualization: ")

# Load and preprocess the image
img = image.load_img(img_path, target_size=(64, 64))
x = image.img_to_array(img)
x = np.expand_dims(x, axis=0) / 255.0 # Add batch dimension and normalize

# ---- Predict and get class index ----
preds = classifier.predict(x)
predicted_class = np.argmax(preds[0])
labels = list(test_generator.class_indices.keys()) # Assuming test_generator is available
predicted_label = labels[predicted_class]

print(f"\n✅ Predicted Class: {predicted_label}")

# ---- Identify last convolutional layer ----
last_conv_layer = None
for layer in reversed(classifier.layers):
    if isinstance(layer, Conv2D):  # Check if the layer is a Conv2D layer
        last_conv_layer = layer
        break

if last_conv_layer is None:
    print("❌ Error: No Conv2D layer found in the model.")
else:
    print(f"🧠 Using last convolutional layer: {last_conv_layer.name}")
```

```python
# ---- Build model to get gradients (using functional API) ----

# Get model input shape safely
if hasattr(classifier, 'input_shape'):
    new_input_shape = classifier.input_shape[1:]
else:
    new_input_shape = classifier.layers[0].input.shape[1:]

new_input = tf.keras.Input(shape=new_input_shape)

# Connect layers sequentially, capturing the last conv layer’s output
x_flow = new_input
conv_layer_output = None
for layer in classifier.layers:
    x_flow = layer(x_flow)
    if layer.name == last_conv_layer.name:
        conv_layer_output = x_flow

if conv_layer_output is None:
    print(f"❌ Could not find last conv layer output for {last_conv_layer.name}")
else:
    grad_model = tf.keras.Model(
        inputs=new_input,
        outputs=[conv_layer_output, x_flow]
    )

    # Compute Grad-CAM
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(x)
        loss = predictions[:, predicted_class]

    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]
    heatmap = tf.reduce_mean(tf.multiply(pooled_grads, conv_outputs), axis=-1)

    heatmap = np.maximum(heatmap, 0)
    if np.max(heatmap) != 0:
        heatmap /= np.max(heatmap)
    else:
        print("⚠️ Heatmap is all zeros. Cannot normalize.")

    # Overlay heatmap
    img_original = cv2.imread(img_path)
    heatmap = cv2.resize(heatmap, (img_original.shape[1], img_original.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    overlay = cv2.addWeighted(img_original, 0.6, heatmap, 0.4, 0)

    # Show results
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img_original, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
    plt.title(f"Grad-CAM (Predicted: {predicted_label})")
    plt.axis('off')
    plt.show()

    print("✅ Grad-CAM visualization generated successfully!")
```
